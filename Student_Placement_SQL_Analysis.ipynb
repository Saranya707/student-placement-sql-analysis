# Student Placement Data Analysis Using SQL

## Problem Statement
Colleges maintain student academic, skill and placement data.
Analyzing this data can help understand placement performance,
salary trends and factors associated with student placements.

## Objective
To use SQL to analyze student placement data and identify
useful placement-related insights.

```python
import sqlite3
import pandas as pd
```

```python
conn = sqlite3.connect('placement.db')

cursor = conn.cursor()

print("Database created successfully!")
```

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS students (
    Student_ID INTEGER PRIMARY KEY,
    Name TEXT,
    Department TEXT,
    CGPA REAL,
    Python_Skill INTEGER,
    SQL_Skill INTEGER,
    Projects INTEGER,
    Placement_Status TEXT,
    Salary_LPA REAL
)
""")

conn.commit()

print("Students table created successfully!")
```

```python
query = """
SELECT *
FROM students;
"""

df_students = pd.read_sql_query(query, conn)

df_students
```

```python
query = """
SELECT *
FROM students
WHERE Placement_Status = 'Placed';
"""

placed_students = pd.read_sql_query(query, conn)

placed_students
```

```python
query = """
SELECT Name, Department, CGPA
FROM students
WHERE CGPA > 8;
"""

high_cgpa_students = pd.read_sql_query(query, conn)

high_cgpa_students
```

```python
query = """
SELECT Name, Department, CGPA
FROM students
ORDER BY CGPA DESC;
"""

result = pd.read_sql_query(query, conn)

result
```

```python
query = """
SELECT Name, Department, CGPA
FROM students
ORDER BY CGPA DESC
LIMIT 3;
"""

top_students = pd.read_sql_query(query, conn)

top_students
```

```python
query = """
SELECT Name, Department, CGPA, Salary_LPA
FROM students
WHERE Placement_Status = 'Placed'
ORDER BY Salary_LPA DESC
LIMIT 5;
"""

top_packages = pd.read_sql_query(query, conn)

top_packages
```

```python
query = """
SELECT COUNT(*) AS Total_Students
FROM students;
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT AVG(CGPA) AS Average_CGPA
FROM students;
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT MAX(Salary_LPA) AS Highest_Package
FROM students;
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT MIN(Salary_LPA) AS Lowest_Placed_Package
FROM students
WHERE Placement_Status = 'Placed';
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT Department,
       AVG(CGPA) AS Average_CGPA
FROM students
GROUP BY Department;
"""

department_cgpa = pd.read_sql_query(query, conn)

department_cgpa
```

```python
query = """
SELECT Department,
       COUNT(*) AS Placed_Students
FROM students
WHERE Placement_Status = 'Placed'
GROUP BY Department;
"""

department_placements = pd.read_sql_query(query, conn)

department_placements
```

```python
query = """
SELECT Department,
       ROUND(AVG(Salary_LPA), 2) AS Average_Salary_LPA
FROM students
WHERE Placement_Status = 'Placed'
GROUP BY Department
ORDER BY Average_Salary_LPA DESC;
"""

department_salary = pd.read_sql_query(query, conn)

department_salary
```

```python
query = """
SELECT Placement_Status,
       COUNT(*) AS Number_of_Students
FROM students
GROUP BY Placement_Status;
"""

placement_summary = pd.read_sql_query(query, conn)

placement_summary
```

```python
query = """
SELECT Department,
       ROUND(AVG(CGPA), 2) AS Average_CGPA
FROM students
GROUP BY Department
HAVING AVG(CGPA) > 8;
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT Name,
       CGPA,
       CASE
           WHEN CGPA >= 9 THEN 'Excellent'
           WHEN CGPA >= 8 THEN 'Good'
           WHEN CGPA >= 7 THEN 'Average'
           ELSE 'Needs Improvement'
       END AS CGPA_Category
FROM students;
"""

cgpa_category = pd.read_sql_query(query, conn)
cgpa_category
```

```python
query = """
SELECT
    ROUND(
        100.0 * SUM(CASE WHEN Placement_Status = 'Placed' THEN 1 ELSE 0 END)
        / COUNT(*), 2
    ) AS Placement_Percentage
FROM students;
"""

placement_percentage = pd.read_sql_query(query, conn)

placement_percentage
```

```python
query = """
SELECT
    Placement_Status,
    ROUND(AVG(Python_Skill), 2) AS Avg_Python_Skill,
    ROUND(AVG(SQL_Skill), 2) AS Avg_SQL_Skill,
    ROUND(AVG(Projects), 2) AS Avg_Projects
FROM students
GROUP BY Placement_Status;
"""

skill_analysis = pd.read_sql_query(query, conn)

skill_analysis
```

```python
query = """
SELECT
    Name,
    Department,
    CGPA,
    Salary_LPA
FROM students
WHERE Placement_Status = 'Placed'
ORDER BY Salary_LPA DESC
LIMIT 1;
"""

highest_paid = pd.read_sql_query(query, conn)

highest_paid
```
